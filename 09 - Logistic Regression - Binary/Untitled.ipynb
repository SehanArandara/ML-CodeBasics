# Logistic Regression 

- we use logistic regression for classification in supervised models
- there are two type of classification in Logistic Regression
- they are single and multiple classification
- we can use sigmoid for single and softmax for multi classification 

```python
import pandas as pd
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('insurance_data.csv')
df.head()
```

```python
plt.scatter(df.age,df.bought_insurance,marker='+',color='red')
```

- 
- THis is the sigmoid or logit function 

```python
x = df.iloc[:,:1]
y= df.iloc[:,1]
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test  =  train_test_split(x,y,test_size=0.2,random_state=10)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
x.shape
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
pred = model.predict(x_test)
pred
```
